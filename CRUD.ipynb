```python
# Ejecuta esta celda en Google Colab
import ipywidgets as widgets
from IPython.display import display, clear_output

# Lista donde se guardan los usuarios temporalmente
usuarios = []

# Elementos de entrada
nombre_input = widgets.Text(placeholder='Nombre', description='Nombre:')
edad_input = widgets.IntText(placeholder='Edad', description='Edad:')
boton_agregar = widgets.Button(description='Agregar', button_style='success')

# Salida para mostrar usuarios
salida = widgets.Output()

# Función para mostrar la lista de usuarios con botones de editar y eliminar
def mostrar_usuarios():
    with salida:
        clear_output()
        if not usuarios:
            print("No hay usuarios registrados.")
        for i, u in enumerate(usuarios):
            print(f"{i}. Nombre: {u['nombre']}, Edad: {u['edad']}")
            btn_editar = widgets.Button(description='Editar', button_style='info')
            btn_eliminar = widgets.Button(description='Eliminar', button_style='danger')

            def editar_callback(b, index=i):
                editar_usuario(index)

            def eliminar_callback(b, index=i):
                eliminar_usuario(index)

            btn_editar.on_click(editar_callback)
            btn_eliminar.on_click(eliminar_callback)
            display(widgets.HBox([btn_editar, btn_eliminar]))

# Función para agregar usuario
def agregar_usuario(b):
    nombre = nombre_input.value.strip()
    edad = edad_input.value
    if nombre and edad > 0:
        usuarios.append({'nombre': nombre, 'edad': edad})
        nombre_input.value = ''
        edad_input.value = 0
        mostrar_usuarios()
    else:
        print("Por favor ingresa un nombre y una edad válidos.")

# Función para editar usuario
def editar_usuario(index):
    usuario = usuarios[index]
    nombre_nuevo = widgets.Text(value=usuario['nombre'], description='Nuevo Nombre:')
    edad_nueva = widgets.IntText(value=usuario['edad'], description='Nueva Edad:')
    btn_guardar = widgets.Button(description='Guardar Cambios', button_style='primary')

    def guardar_cambios(b):
        nuevo_nombre = nombre_nuevo.value.strip()
        nueva_edad = edad_nueva.value
        if nuevo_nombre and nueva_edad > 0:
            usuarios[index] = {'nombre': nuevo_nombre, 'edad': nueva_edad}
            mostrar_usuarios()

    display(widgets.VBox([nombre_nuevo, edad_nueva, btn_guardar]))
    btn_guardar.on_click(guardar_cambios)

# Función para eliminar usuario
def eliminar_usuario(index):
    usuarios.pop(index)
    mostrar_usuarios()

# Conectar botón agregar con su función
boton_agregar.on_click(agregar_usuario)

# Mostrar la interfaz
display(widgets.VBox([nombre_input, edad_input, boton_agregar, salida]))
mostrar_usuarios()
```
